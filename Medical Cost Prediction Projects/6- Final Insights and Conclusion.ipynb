# **Key Insights**

##### **1. Importance of Data Transformation**

Applying log transformation consistently improved performance across multiple models (Linear, Polynomial, SVR, Decision Tree, and Random Forest).
This confirms that the target variable was skewed, and transforming it helped models learn more stable and meaningful patterns.

##### **2. Ensemble Models Perform Best**

Tree based ensemble methods (especially Random Forest) outperformed single models and kernel-based approaches due to their ability to:

- Reduce variance
- Capture complex non linear relationships
- Improve generalization stability

##### **3. Bias–Variance Tradeoff is Critical**

- Simple models (Linear Regression) underfitted the data
- Highly complex models (deep trees or unconstrained models) overfitted
- The best performance came from moderate complexity (max_depth=5)

##### **4. SVR is Sensitive but Competitive**

SVR with RBF kernel performed strongly and showed that kernel choice significantly impacts performance. However, it was slightly less stable compared to ensemble methods.

##### **5. Decision Tree Alone is Not Enough**

Decision Trees improved after tuning but still showed instability compared to Random Forest, confirming that ensembles are more reliable for structured tabular data.

# **Final Conclusion: Medical Cost Prediction Experiment**

Across all experimented models (Linear Regression, Polynomial Regression, SVR, Decision Tree, and Random Forest), the results clearly show that model performance is strongly influenced by both data transformation and model complexity control.

The best overall performance was achieved using Random Forest with log transformation (max_depth=5), reaching a test R² of 0.8561, making it the most accurate and stable model in this study.

In medical cost prediction, data preprocessing (especially log transformation) is as important as model selection, and Random Forest provides the best balance between accuracy and generalization for this dataset.
