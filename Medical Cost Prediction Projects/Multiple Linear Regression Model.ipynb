# **Data preprocessing**

## **Data Cleaning Check Template**
This template is designed to quickly assess the quality of any dataset before building machine learning models or performing analysis.

It provides a structured overview of the dataset by checking for common data issues such as:

- Missing values

- Duplicate rows

- Incorrect data types

- Outliers

- Distribution of numerical features

- Categorical feature consistency

**What This Template Does**

- Displays basic dataset information (shape, data types)

- Identifies missing values and duplicates

- Summarizes numerical and categorical features

- Detects potential outliers using the IQR method

- Highlights columns with low unique values for quick inspection

How to Use

1. Load your dataset using Pandas  

2. Call the function:

````python
def data_quality_report(df):

    print("DATA QUALITY REPORT")
    
    # Print a separator line for better readability
    
    print("=" * 50)
    print("BASIC INFO")
    print("=" * 50)
    
    # Show general information about the dataset (columns, data types, non-null values)
    print(df.info())
    
    # Show number of rows and columns
    print("\n" + "=" * 50)
    print("SHAPE OF DATA")
    print("=" * 50)
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
    
    # Check for missing (null) values in each column
    print("\n" + "=" * 50)
    print("MISSING VALUES")
    print("=" * 50)
    missing = df.isnull().sum()
    
    # Display only columns that have missing values
    print(missing[missing > 0])
    
    # Check for duplicate rows
    print("\n" + "=" * 50)
    print("DUPLICATES")
    print("=" * 50)
    print(f"Duplicate rows: {df.duplicated().sum()}")
    
    # Display data types of each column
    print("\n" + "=" * 50)
    print("DATA TYPES")
    print("=" * 50)
    print(df.dtypes)
    
    # Summary statistics for numerical columns (mean, std, min, max, etc.)
    print("\n" + "=" * 50)
    print("NUMERICAL SUMMARY")
    print("=" * 50)
    print(df.describe())
    
    # Summary for categorical (object) columns
    print("\n" + "=" * 50)
    print("CATEGORICAL SUMMARY")
    print("=" * 50)
    print(df.describe(include=['object']))
    
    # Show unique values for columns with low number of distinct values
    # Useful for detecting categories, errors, or inconsistencies
    print("\n" + "=" * 50)
    print("UNIQUE VALUES (LOW CARDINALITY)")
    print("=" * 50)
    for col in df.columns:
        if df[col].nunique() < 10:  # Only show columns with few unique values
            print(f"{col}: {df[col].unique()}")
            
    # correlation
    print("\n" + "=" * 50)
    print("CORRELATION MATRIX")
    print("=" * 50)
    print(df.corr(numeric_only=True))
    
    # Detect outliers using the IQR (Interquartile Range) method
    print("\n" + "=" * 50)
    print("OUTLIERS CHECK (IQR METHOD)")
    print("=" * 50)
    
    # Loop through only numerical columns
    for col in df.select_dtypes(include=np.number).columns:
        Q1 = df[col].quantile(0.25)  # 25th percentile
        Q3 = df[col].quantile(0.75)  # 75th percentile
        IQR = Q3 - Q1  # Interquartile range
        
        # Count rows that fall outside the normal range
        outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
        print(f"{col}: {len(outliers)} outliers")
````

### Log Transformation vs. Feature Scaling

In data preprocessing, **Log Transformation** fixes the **shape** of your distribution, while **Feature Scaling** fixes the **scale/magnitude** of your features.

---

#### 1. Log Transformation
**When to use:** Use this when your data is **right-skewed** (a long tail of large values) or has **heteroscedasticity** (variance increases with the mean).

* **Key Purpose:** It compresses the range of large values and expands the range of small values to make the distribution more **Gaussian (Normal)**.
* **Common Use Cases:**
    * Financial data (Income, House Prices).
    * Biological counts or population data.
    * When the relationship between variables is **multiplicative** (e.g., exponential growth).
* **Effect on Outliers:** It significantly reduces the relative distance of extreme outliers.



---

#### 2. Feature Scaling
**When to use:** Use this when features have **different units** (e.g., Age in years vs. Annual Income in EGP).

* **Key Purpose:** It ensures that no single feature dominates the model simply because it has a larger numerical range.
* **Techniques:**
    * **Standardization (Z-score):** Centers data around a mean of 0 with a standard deviation of 1.
    * **Normalization (Min-Max):** Rescales data to a fixed range, usually $[0, 1]$.
* **Required for:**
    * **Distance-based algorithms:** KNN, K-Means, SVM (where distances between points matter).
    * **Gradient-Descent algorithms:** Linear/Logistic Regression, Neural Networks (converges much faster).
    * **Dimensionality Reduction:** PCA (which looks for maximum variance).



---

#### Summary Table

| Feature | Log Transformation | Feature Scaling |
| :--- | :--- | :--- |
| **Primary Goal** | Change the **distribution shape**. | Change the **range/magnitude**. |
| **Main Problem** | Skewness / Outliers. | Unit mismatch / Slow convergence. |
| **Model Type** | Useful for almost all models if data is skewed. | Essential for Distance & Gradient models. |
| **Note** | Data must be positive ($x > 0$). | Can be applied to any numerical data. |

**Pro Tip:** For **Tree-based models** (Decision Trees, Random Forests), feature scaling is generally **not required** because the model splits based on rank, but Log Transformation can still help if the *target* variable is highly skewed.

# **Important Insights**


1. Log transformation significantly improved model performance, increasing the R² score from 0.7530 to 0.7716. This indicates that the target variable (charges) is right-skewed and benefits from a more normally distributed target.
2. Feature scaling had no effect on model performance. This is expected because Multiple Linear Regression is not sensitive to feature scale in terms of prediction accuracy.
3. After applying log transformation, no further improvement was observed. This suggests that preprocessing alone cannot overcome the inherent limitations of the model.
4. The main performance gain came from addressing data distribution, not from modifying feature magnitudes.

### Why Multiple Linear Regression is Limited in This Problem

Assumption of Linear Relationships
1. Multiple Linear Regression assumes a strictly linear relationship between input features and the target variable.

    However, in this dataset:

    - The effect of BMI on charges is not purely linear
    - The impact of smoking creates sharp, non-linear jumps in cost

These patterns cannot be fully captured by a linear model.
