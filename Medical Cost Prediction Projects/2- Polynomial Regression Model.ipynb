# **Data preprocessing**

## **Data Cleaning Check Template**
This template is designed to quickly assess the quality of any dataset before building machine learning models or performing analysis.

It provides a structured overview of the dataset by checking for common data issues such as:

- Missing values

- Duplicate rows

- Incorrect data types

- Outliers

- Distribution of numerical features

- Categorical feature consistency

**What This Template Does**

- Displays basic dataset information (shape, data types)

- Identifies missing values and duplicates

- Summarizes numerical and categorical features

- Detects potential outliers using the IQR method

- Highlights columns with low unique values for quick inspection

How to Use

1. Load your dataset using Pandas  

2. Call the function:

````python
def data_quality_report(df):

    print("DATA QUALITY REPORT")
    
    # Print a separator line for better readability
    
    print("=" * 50)
    print("BASIC INFO")
    print("=" * 50)
    
    # Show general information about the dataset (columns, data types, non-null values)
    print(df.info())
    
    # Show number of rows and columns
    print("\n" + "=" * 50)
    print("SHAPE OF DATA")
    print("=" * 50)
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
    
    # Check for missing (null) values in each column
    print("\n" + "=" * 50)
    print("MISSING VALUES")
    print("=" * 50)
    missing = df.isnull().sum()
    
    # Display only columns that have missing values
    print(missing[missing > 0])
    
    # Check for duplicate rows
    print("\n" + "=" * 50)
    print("DUPLICATES")
    print("=" * 50)
    print(f"Duplicate rows: {df.duplicated().sum()}")
    
    # Display data types of each column
    print("\n" + "=" * 50)
    print("DATA TYPES")
    print("=" * 50)
    print(df.dtypes)
    
    # Summary statistics for numerical columns (mean, std, min, max, etc.)
    print("\n" + "=" * 50)
    print("NUMERICAL SUMMARY")
    print("=" * 50)
    print(df.describe())
    
    # Summary for categorical (object) columns
    print("\n" + "=" * 50)
    print("CATEGORICAL SUMMARY")
    print("=" * 50)
    print(df.describe(include=['object']))
    
    # Show unique values for columns with low number of distinct values
    # Useful for detecting categories, errors, or inconsistencies
    print("\n" + "=" * 50)
    print("UNIQUE VALUES (LOW CARDINALITY)")
    print("=" * 50)
    for col in df.columns:
        if df[col].nunique() < 10:  # Only show columns with few unique values
            print(f"{col}: {df[col].unique()}")
            
    # correlation
    print("\n" + "=" * 50)
    print("CORRELATION MATRIX")
    print("=" * 50)
    print(df.corr(numeric_only=True))
    
    # Detect outliers using the IQR (Interquartile Range) method
    print("\n" + "=" * 50)
    print("OUTLIERS CHECK (IQR METHOD)")
    print("=" * 50)
    
    # Loop through only numerical columns
    for col in df.select_dtypes(include=np.number).columns:
        Q1 = df[col].quantile(0.25)  # 25th percentile
        Q3 = df[col].quantile(0.75)  # 75th percentile
        IQR = Q3 - Q1  # Interquartile range
        
        # Count rows that fall outside the normal range
        outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
        print(f"{col}: {len(outliers)} outliers")
````

### Log Transformation vs. Feature Scaling

In data preprocessing, **Log Transformation** fixes the **shape** of your distribution, while **Feature Scaling** fixes the **scale/magnitude** of your features.

---

#### 1. Log Transformation
**When to use:** Use this when your data is **right-skewed** (a long tail of large values) or has **heteroscedasticity** (variance increases with the mean).

* **Key Purpose:** It compresses the range of large values and expands the range of small values to make the distribution more **Gaussian (Normal)**.
* **Common Use Cases:**
    * Financial data (Income, House Prices).
    * Biological counts or population data.
    * When the relationship between variables is **multiplicative** (e.g., exponential growth).
* **Effect on Outliers:** It significantly reduces the relative distance of extreme outliers.



---

#### 2. Feature Scaling
**When to use:** Use this when features have **different units** (e.g., Age in years vs. Annual Income in EGP).

* **Key Purpose:** It ensures that no single feature dominates the model simply because it has a larger numerical range.
* **Techniques:**
    * **Standardization (Z-score):** Centers data around a mean of 0 with a standard deviation of 1.
    * **Normalization (Min-Max):** Rescales data to a fixed range, usually $[0, 1]$.
* **Required for:**
    * **Distance-based algorithms:** KNN, K-Means, SVM (where distances between points matter).
    * **Gradient-Descent algorithms:** Linear/Logistic Regression, Neural Networks (converges much faster).
    * **Dimensionality Reduction:** PCA (which looks for maximum variance).



---

#### Summary Table

| Feature | Log Transformation | Feature Scaling |
| :--- | :--- | :--- |
| **Primary Goal** | Change the **distribution shape**. | Change the **range/magnitude**. |
| **Main Problem** | Skewness / Outliers. | Unit mismatch / Slow convergence. |
| **Model Type** | Useful for almost all models if data is skewed. | Essential for Distance & Gradient models. |
| **Note** | Data must be positive ($x > 0$). | Can be applied to any numerical data. |

**Pro Tip:** For **Tree-based models** (Decision Trees, Random Forests), feature scaling is generally **not required** because the model splits based on rank, but Log Transformation can still help if the *target* variable is highly skewed.

# **Encoding categorical data**

## Encoding the Indpendent Variable Using One Hot Encoding

Note: I explained One-Hot Encoding earlier.

If you missed it or want a deeper understanding of how it works, you can find it here: [Go to One-Hot Encoding Section](https://github.com/Hazem1695/ml-concept-briefs)

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
# One-hot encode the 'color' column (index 2) while keeping other features
# Initialize ColumnTransformer to handle categorical data
# [2] targets the 'color' column (index 2)
ct = ColumnTransformer(transformers = [('encoder', OneHotEncoder(sparse_output=False),[1,4,5])], remainder = 'passthrough')
# Execute transformation and convert to NumPy ndarray for model training
X = np.array(ct.fit_transform(X))
````

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train , y_test = train_test_split(X, y_log , test_size = 0.2, random_state = 0)
````

# **Building Polynomial Regression Model**

## Training the Polynomial Regression model on the Training set

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly_reg = PolynomialFeatures(degree = 2)
X_poly = poly_reg.fit_transform(X_train)

regressor = LinearRegression()
regressor.fit(X_poly, y_train)
````

# **Preticting the Test set results**

````python
y_pred_log = regressor.predict(poly_reg.transform(X_test)).reshape(-1,1)
y_pred = np.exp(y_pred_log)
np.set_printoptions(precision=2,suppress=True)
print(np.concatenate((y_pred.reshape(len(y_pred), 1), np.exp(y_test).reshape(len(y_test), 1)), axis = 1))
````

# **Evaluating the Model Performance**

````python
from sklearn.metrics import r2_score
r2_score(y_test, y_pred_log)
````

````python
y_trained = regressor.predict(X_poly)
r2_score(y_train, y_trained)
````

# **Technical Insights**

### 1. Target Transformation Matters More Than Feature Scaling

The improvement from 0.8349 → 0.8495 shows that:

- The model was struggling with target distribution, not feature representation
- Applying log reduced:
    - skewness
    - variance instability (heteroscedasticity)

    This indicates a data-centric improvement, not just model tuning

### 2. Feature Scaling Had No Effect

Feature scaling produced no measurable improvement

This is expected because:

- Polynomial Regression is built on Linear Regression
- It is not sensitive to feature scale when:
    - no regularization is used
    - closed-form solution is applied

### 3. Model vs Data: Who solved the problem?

- Polynomial features → handled non-linearity in X
- Log transformation → handled distribution issues in y

Improvement came from better problem formulation, not increasing model complexity

Note: Fix the data before ove engineering the model

### 4. Diminishing Returns from Model Complexity
    
- Polynomial regression already captured most patterns
- Log transformation gave a targeted improvement
- Further preprocessing (scaling) → no gain

# **Conclusion**

This experiment highlights that data-centric improvements can be as impactful as model selection. Across both Linear Regression and Polynomial Regression, applying a log transformation to the target variable consistently improved performance, indicating the presence of skewness and heteroscedasticity. By stabilizing variance and reducing the influence of extreme values, log transformation led to more reliable and generalizable predictions.

While Polynomial Regression significantly outperformed Linear Regression by capturing non-linear relationships in the data, the combination of model choice and proper target transformation yielded the best results. In contrast, feature scaling showed no measurable impact, which aligns with the theoretical properties of linear-based models without regularization.

Additionally, the model demonstrated strong generalization behavior, with test performance slightly exceeding training performance. This suggests that the model avoids overfitting and benefits from stable learning dynamics.

Overall, the results emphasize that understanding data distribution and aligning preprocessing techniques with model behavior is more effective than blindly applying standard techniques. The most effective pipeline combines Polynomial Regression with log transformation, achieving a balance between model expressiveness and data stability.
